```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Загружаем данные
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

print(f"Размер данных: {df.shape}")
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing': missing, 'Percent': missing_pct})
missing_df[missing_df['Missing'] > 0].sort_values('Missing', ascending=False)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Sex', y='Survived', data=df, ci=None)
plt.title('Доля выживших по полу')
plt.ylabel('Доля выживших')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Sex', y='Survived', data=df, ci=None)
plt.title('Доля выживших по полу')
plt.ylabel('Доля выживших')
plt.show()
```

```python
fig, axes = plt.subplots(1,2, figsize=(14,5))
sns.histplot(data=df, x='Age', hue='Survived', kde=True, ax=axes[0])
axes[0].set_title('Распределение возраста по выживаемости')
sns.countplot(x='Pclass', hue='Survived', data=df, ax=axes[1])
axes[1].set_title('Выживаемость по классу билета')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
numeric_cols = ['Survived','Pclass','Age','SibSp','Parch','Fare']
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Корреляция числовых признаков')
plt.show()
```
